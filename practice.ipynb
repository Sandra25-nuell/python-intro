```python
data_set ={
    "law" :{"Sandra", "Ada", "Joy"},
    "Bio" :{"Maya", "Ola", "Mercy"}
}

data_set["law"].add("king")
data_set["law"].add("Alice")

#get intersection and difference of course a and b
student_in_both = data_set["law"].intersection(data_set["Bio"])
only_law = data_set["law"].difference(data_set["Bio"])

print(data_set)
print(student_in_both)
print(only_law)

# count students per course
print("=" * 20)
print("solution 4")

print("=" * 20)

for course, students in data_set.items():
    print(f"{course} - {len(students)}")


#find the course with the highest student count
highest_count = 0
winning_course = ""

for course, students in data_set.items():
    if len(students) > highest_count:
        winning_course = course

print(winning_course)
```
